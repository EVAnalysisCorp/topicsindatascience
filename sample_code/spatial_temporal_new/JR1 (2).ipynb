```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt

# Load the world map from GeoPandas datasets
world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))

# Merge your data with world map data on country name
# Assuming 'geo' has a 'Country name' column that matches with 'name' in the world dataset
merged = world.merge(geo, how="left", left_on="name", right_on="Country name")

# Plot a choropleth map for Log GDP per capita
fig, ax = plt.subplots(1, 1, figsize=(15, 10))
merged.boundary.plot(ax=ax)
merged.plot(column='Log GDP per capita', ax=ax, legend=True, 
            cmap='OrRd', missing_kwds={"color": "lightgrey"})

plt.title("Log GDP per capita by Country")
plt.show()
```

```python
correlation_data = geo[['Country name', 'Healthy life expectancy at birth', 'Freedom to make life choices']]

# Merge the correlation data with geospatial data
merged_corr = world.merge(correlation_data, how="left", left_on="name", right_on="Country name")

# Plot the correlation map
fig, ax = plt.subplots(1, 1, figsize=(15, 10))
merged_corr.plot(column='Healthy life expectancy at birth', ax=ax, legend=True, cmap='Blues', missing_kwds={"color": "lightgrey"})
plt.title("Healthy Life Expectancy by Country")
plt.show()
```

```python
import pycountry_convert as pc
import pycountry
import pandas as pd

# Function to get the continent from a country name
def country_to_continent(country_name):
    try:
        country_alpha2 = pycountry.countries.lookup(country_name).alpha_2
        continent_code = pc.country_alpha2_to_continent_code(country_alpha2)
        continents = {
            'AF': 'Africa',
            'NA': 'North America',
            'SA': 'South America',
            'AS': 'Asia',
            'OC': 'Oceania',
            'EU': 'Europe',
            'AN': 'Antarctica'
        }
        return continents[continent_code]
    except:
        return None  # Return None if the country name is not recognized

# Apply the function to your 'Country name' column
geo['Continent'] = geo['Country name'].apply(country_to_continent)

# Show the updated dataframe with the new 'Continent' column
print(geo.head())
```

```python
# Group by 'Continent' and calculate the mean Life Ladder (happiness) score
happiness_by_continent = geo.groupby('Continent')['Life Ladder'].mean().reset_index()

# Show the resulting DataFrame
print(happiness_by_continent)
```

```python
# Sort by Life Ladder (happiness) score and get the top 5 and bottom 5 countries
top_5_happiest = geo.nlargest(5, 'Life Ladder')
bottom_5_least_happy = geo.nsmallest(5, 'Life Ladder')

# Concatenate the two dataframes
top_bottom_10 = pd.concat([top_5_happiest, bottom_5_least_happy])

# Check the result
print(top_bottom_10[['Country name', 'Life Ladder']])

gdf_top_bottom = gpd.GeoDataFrame(top_bottom_10, geometry=gpd.points_from_xy(top_bottom_10.Longitude, top_bottom_10.Latitude))

# Set up the plot
fig, ax = plt.subplots(figsize=(15, 10))

# Plot the world map
world.plot(ax=ax, color='lightgrey')

# Plot green dots for the top 5 happiest countries
gdf_top_bottom[gdf_top_bottom['Life Ladder'] >= top_5_happiest['Life Ladder'].min()].plot(ax=ax, color='green', marker='o', markersize=100, label='Top 5 Happiest')

# Plot red dots for the bottom 5 least happy countries
gdf_top_bottom[gdf_top_bottom['Life Ladder'] <= bottom_5_least_happy['Life Ladder'].max()].plot(ax=ax, color='red', marker='o', markersize=100, label='Bottom 5 Least Happy')

# Add legend and title
plt.legend()
plt.title('Top 5 Happiest and Bottom 5 Least Happy Countries', fontsize=16)

# Show the map
plt.show()
```

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

# Define the top and bottom 5 happiest countries with their corresponding latitudes and longitudes
data = {
    'Country name': [
        'Finland', 'Denmark', 'Iceland', 'Israel', 'Netherlands',  # Top 5
        'Honduras', 'Yemen', 'India', 'Syria', 'Zimbabwe'          # Bottom 5
    ],
    'Life Ladder': [
        7.8, 7.6, 7.5, 7.5, 7.4,  # Top 5 happiness scores
        3.9, 3.8, 3.7, 3.6, 3.5   # Bottom 5 happiness scores
    ],
    'Latitude': [
        61.9241, 56.2639, 64.9631, 31.0461, 52.1326,  # Top 5 latitudes
        13.4978, 15.5522, 20.5937, 34.8021, -19.0154   # Bottom 5 latitudes
    ],
    'Longitude': [
        25.7482, 9.5018, -18.0105, 34.8516, 5.3872,  # Top 5 longitudes
        -86.2419, 48.5164, 78.9629, 38.9968, 29.1549   # Bottom 5 longitudes
    ]
}

# Create a DataFrame
top_bottom_10 = pd.DataFrame(data)

# Create GeoDataFrame using manually inputted coordinates
gdf_top_bottom = gpd.GeoDataFrame(
    top_bottom_10, 
    geometry=gpd.points_from_xy(top_bottom_10.Longitude, top_bottom_10.Latitude)
)

# Create a base map using a simple plot
world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))

# Set up the plot
fig, ax = plt.subplots(figsize=(15, 10))

# Plot the world map
world.plot(ax=ax, color='lightgrey')

# Plot green dots for the top 5 happiest countries
gdf_top_bottom.iloc[:5].plot(ax=ax, color='green', marker='o', markersize=100, label='Top 5 Happiest')

# Plot red dots for the bottom 5 least happy countries
gdf_top_bottom.iloc[5:].plot(ax=ax, color='red', marker='o', markersize=100, label='Bottom 5 Least Happy')

# Add legend and title
plt.legend()
plt.title('Top 5 Happiest and Bottom 5 Least Happy Countries', fontsize=16)

# Show the map
plt.show()
```
